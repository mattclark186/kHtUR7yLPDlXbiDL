```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
from nltk.stem import PorterStemmer
```

```python
data = pd.read_csv('../data/potential_talents_cleaned.csv')
```

```python
data.head()
```

```python
# choosing sentence transformer model to use
model = SentenceTransformer('all-MiniLM-L6-v2')

# creating embeddings of the words in the job titles
job_embeddings = model.encode(data['job_title'].to_list())

# creating embeddings of the target job title
stemmer = PorterStemmer()
target_title = 'aspiring hr'
target_stem = ' '.join(
    stemmer.stem(word)
    for word in target_title.split()
    )
target_embedding = model.encode([target_stem])

# finding the similarity between each job title and the target
similarities = cosine_similarity(
    target_embedding,
    job_embeddings
).flatten()

# adding the similiarities to the dataframe
data['similarity'] = similarities
data = data.sort_values(by='similarity', ascending=False).reset_index(drop=True)

data
```

```python
# choosing sentence transformer model to use
model = SentenceTransformer('all-MiniLM-L6-v2')

# creating embeddings of the words in the job titles
job_embeddings = model.encode(data['job_title_stem'].to_list())

# creating embeddings of the target job title
stemmer = PorterStemmer()
target_title = 'aspiring hr'
target_stem = ' '.join(
    stemmer.stem(word)
    for word in target_title.split()
    )
target_embedding = model.encode([target_stem])

# finding the similarity between each job title and the target
similarities = cosine_similarity(
    target_embedding,
    job_embeddings
).flatten()

# adding the similiarities to the dataframe
data['similarity_stem'] = similarities
data = data.sort_values(by='similarity_stem', ascending=False).reset_index(drop=True)

data
```
